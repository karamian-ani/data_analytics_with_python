```python
X = df.drop(columns=["churn"])
y = df["churn"]
```

```python
X
```

```python
education_order = ["High School", "Bachelor", "Master", "PhD"]

df["education_level"] = pd.Categorical(
    df["education_level"],
    categories=education_order,
    ordered=True
)

education_dummies = pd.get_dummies(
    df["education_level"],
    prefix="education",
    drop_first=True,
    dtype=int
)

education_dummies.head()
```

```python
X_encoded = pd.concat([X,education_dummies],axis=1)
X_encoded.head()
```

```python
X = X_encoded
#del X_encoded
```

```python
def test():
    return ('X_train', 'X_test', 'y_train', 'y_test' )
```

```python
x1, x2, x3, x4 = test()
print (x1)
print (x2)
print (x3)
print (x4)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
X_train.head()
```

```python
train_size = len(X_train)
test_size = len(X_test)

train_churn_rate = y_train.mean()
test_churn_rate = y_test.mean()

print("Training rows:", train_size)
print("Test rows:", test_size)
print("Training churn rate:", round(train_churn_rate, 3))
print("Test churn rate:", round(test_churn_rate, 3))
```

```python
# for fixing the following model code error
X_train.select_dtypes(include="object").columns
```

```python
# for fixing the following model code error
print(X_train.select_dtypes(include="object").columns)
print(X_train.select_dtypes(include="object").nunique())
```

```python
# for fixing the following model code error
X_train, X_test = X_train.align(X_test, join="left", axis=1, fill_value=0)
```

```python
# for fixing the following model code error
X_train = pd.get_dummies(X_train, columns=["education_level"])
X_test = pd.get_dummies(X_test, columns=["education_level"])
```

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)
```

```python
probs = model.predict_proba(X_test)
# [1-p, p]
probs = probs[:,1]
```

```python
threshold = 0.5
preds = (probs >= threshold).astype(int)
preds[:5]
```

```python
prediction_results = X_test.copy()
prediction_results['actual_churn'] = y_test.values
prediction_results['prediction_probs'] = probs
prediction_results['prediction_churn'] = preds
```

```python
prediction_results.head()
```

**Outcome	Meaning**
- **True Positive**	Customer churned and model predicted churn
- **False Positive**	Customer did not churn but model predicted churn
- **False Negative**	Customer churned but model missed it
- **True Negative**	Customer did not churn and model predicted no churn

```python
prediction_results[['actual_churn','prediction_churn']].value_counts()
```

```python
cm = confusion_matrix(y_test, preds)

tn, fp, fn, tp = cm.ravel()

cm
```

```python
accuracy = accuracy_score(y_test, preds) #(103+50)/200
precision = precision_score(y_test, preds) #50/(50+24)
recall = recall_score(y_test, preds) #50/(50+23)

print("Accuracy:", round(accuracy, 3))
print("Precision:", round(precision, 3))
print("Recall:", round(recall, 3))
```

```python
cm = confusion_matrix(y_test,preds,normalize='all')
cm
```

```python
cm = confusion_matrix(y_test,preds,normalize='true')
cm
```

```python
retention_cost = 10
retention_benefit = 50 #COHORT Analysis

revenue_from_saved_customers = tp * (retention_benefit - retention_cost)
cost_from_unnecessary_contacts = fp * retention_cost

net_profit = revenue_from_saved_customers - cost_from_unnecessary_contacts

print("Revenue from saved customers:", revenue_from_saved_customers)
print("Cost from unnecessary contacts:", cost_from_unnecessary_contacts)
print("Net profit:", net_profit)
```
